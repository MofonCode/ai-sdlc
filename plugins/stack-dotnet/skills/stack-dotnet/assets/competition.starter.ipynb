# {{SOLUTION}} — exploration

A scratchpad for this competition, on the .NET Interactive kernel. It ships as headings
and nothing else, deliberately: a notebook has no project file, so `Directory.Build.props`
never reaches it and there are **no analyzers, no `EnforceCodeStyleInBuild` and no
warnings-as-errors** here. Anything shipped in this file would be code the gate the rest of
the solution relies on had never seen.

> **Nothing graduates from this notebook to a deliverable by being copied.** It moves into
> `src/{{SOLUTION}}.Features/`, compiles under the same analyzers as everything else, and
> gets tests in `tests/{{SOLUTION}}.UnitTests/`. Until then it is a sketch, regardless of
> how finished it looks.

Open it in Polyglot Notebooks (VS Code), or in any Jupyter frontend with the .NET
Interactive kernel registered — it is an ordinary `.ipynb` either way.
